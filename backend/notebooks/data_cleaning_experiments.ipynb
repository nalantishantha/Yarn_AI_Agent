# Data Cleaning Experiments
Here we test simple, step-by-step methods to extract numerical values from our string columns.

```python
import pandas as pd
```

### 1. Load the Raw Data

```python
csv_path = '../../dataset/Yarn_Supplier_Database_Clean.csv'
df = pd.read_csv(csv_path)
print(f"Loaded {len(df)} rows.")
```

```python
df.info()
```

### 2. Clean Lead Time (`Manufacture LT(Week)`)
We will extract the numbers, treat the first number as MIN, the second as MAX, and multiply by 7 to get days.

```python
# Find all numbers in the string column
extracted_lt = df['Manufacture LT(Week)'].astype(str).str.findall(r'\d+')

# Get the first number (MIN week)
lt_min_weeks = extracted_lt.str[0].astype(float)

# Get the second number (MAX week). If it doesn't exist, use the MIN week again.
lt_max_weeks = extracted_lt.str[1].astype(float).fillna(lt_min_weeks)

# Multiply by 7 to get days
df['lt_min_days'] = lt_min_weeks * 7
df['lt_max_days'] = lt_max_weeks * 7

# Check the results
df[['Manufacture LT(Week)', 'lt_min_days', 'lt_max_days']].dropna().head()
```

### 3. Clean MOQ (`MOQ(Kg)`)
We will do the same thing: extract the numbers for MIN and MAX MOQ.

```python
# Find all numbers (including decimals) in the string column
extracted_moq = df['MOQ(Kg)'].astype(str).str.findall(r'\d+\.?\d*')

# Get the first number (MIN MOQ)
df['moq_min'] = extracted_moq.str[0].astype(float)

# Get the second number (MAX MOQ). If it doesn't exist, use the MIN MOQ again.
df['moq_max'] = extracted_moq.str[1].astype(float).fillna(df['moq_min'])

# Check the results
df[['MOQ(Kg)', 'moq_min', 'moq_max']].dropna().head()
```
